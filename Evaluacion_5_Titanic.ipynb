## Cargar los datos

Se lee el archivo CSV con los datos del Titanic.

```r
# Leer el archivo CSV
DATOS <- read.csv("titanic.csv")

# Ver las primeras 10 filas para entender la estructura
head(DATOS, 10)
```

## Pregunta 1
**Muestre una tabla donde se visualice la Edad promedio por Puerto de embarque. ¿Cuál fue el puerto donde hubo el mayor promedio de edad?**

```r
# Convertir Embarked a factor para agrupar correctamente
DATOS$Embarked <- as.factor(DATOS$Embarked)

# Agrupar por puerto y calcular el promedio de edad
DATOS %>%
  group_by(Embarked) %>%
  summarise(Edad_Promedio = mean(Age, na.rm = TRUE))
```

**Respuesta:** El puerto con la edad promedio mayor es el puerto **C (Cherbourg)**, con un promedio de aproximadamente 30.81 años. Le sigue el puerto S (Southampton) con 29.45 años y por último el puerto Q (Queenstown) con 28.09 años.

## Pregunta 2
**En un mismo cuadro, grafique los histogramas de edad por estado de sobrevivencia (Survived). Comparando estos dos histogramas, ¿qué puede concluir?**

```r
# Convertir Survived a factor
DATOS$Survived <- as.factor(DATOS$Survived)

# Separar los datos en sobrevivientes y fallecidos
SURVIVED <- DATOS %>% filter(Survived == 1)
DECEASED <- DATOS %>% filter(Survived == 0)

# Histograma de los SOBREVIVIENTES
SURVIVED %>% ggplot(aes(x = Age)) +
  geom_histogram(fill = "green", color = "black", bins = 25) +
  ggtitle("Histograma de Edad - Sobrevivientes") +
  labs(x = "Edad (años)", y = "Frecuencia") +
  theme_minimal()
```

```r
# Histograma de los FALLECIDOS
DECEASED %>% ggplot(aes(x = Age)) +
  geom_histogram(fill = "red", color = "black", bins = 25) +
  ggtitle("Histograma de Edad - Fallecidos") +
  labs(x = "Edad (años)", y = "Frecuencia") +
  theme_minimal()
```

**Respuesta:** Comparando ambos histogramas, se observa que la supervivencia no depende significativamente de la edad, ya que ambas gráficas tienen una forma similar con los valores más altos concentrados alrededor de los 25-35 años. Sin embargo, se puede notar que las personas entre 10 y 20 años tienden a tener mayor proporción de sobrevivencia, y los niños pequeños (0-10 años) también muestran mayor tasa de supervivencia.

## Pregunta 3
**Grafique los boxplots de edad por sexo. Comparando ambos boxplots, ¿cuál es la conclusión?**

```r
# Boxplot de Edad separado por Sexo
ggplot(DATOS, aes(x = Sex, y = Age, fill = Sex)) +
  geom_boxplot() +
  ggtitle("Boxplot de Edad por Sexo") +
  labs(x = "Sexo", y = "Edad") +
  theme_minimal()
```

**Respuesta:** La mediana de edad para los hombres es ligeramente mayor que la de las mujeres. Además, los hombres presentan un rango intercuartílico más amplio, lo que indica mayor dispersión en las edades. Los hombres a bordo tienden a tener una edad mayor en comparación con las mujeres.

## Pregunta 5
**Grafique un diagrama de barras para sobrevivencia. ¿Cuántas personas sobrevivieron, y cuántas fallecieron?**

```r
# Ver la tabla de frecuencias
table(DATOS$Survived)

# Diagrama de barras
ggplot(DATOS, aes(x = factor(Survived))) +
  geom_bar(fill = c("tomato", "steelblue")) +
  labs(title = "Sobrevivencia en el Titanic",
       x = "Sobrevivió (0 = No, 1 = Sí)",
       y = "Cantidad de personas") +
  theme_minimal()
```

**Respuesta:** Fallecieron **549** personas y sobrevivieron **342** personas. La mayoría de los pasajeros (aproximadamente el 61.6%) no sobrevivió al hundimiento.

---
## Pregunta 6
**Diagrama de barras para sobrevivencia, diferenciado por sexo. ¿Qué puede concluir?**

```r
# Diagrama de barras de sobrevivencia por sexo
ggplot(DATOS, aes(x = Sex, fill = factor(Survived))) +
  geom_bar(position = "dodge") +
  labs(title = "Sobrevivencia por Sexo",
       x = "Sexo",
       y = "Cantidad de personas",
       fill = "Sobrevivió") +
  theme_minimal()
```

**Respuesta:** La cantidad de mujeres que sobrevivieron es notablemente mayor que la cantidad que no sobrevivió, mientras que para los hombres ocurre lo contrario: la gran mayoría de los hombres fallecieron. Esto refleja la política de "mujeres y niños primero" durante la evacuación del Titanic. Las mujeres tuvieron una tasa de supervivencia mucho mayor que los hombres.
